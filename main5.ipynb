```python
import numpy as np
```

```python
X=np.array([[40,5],
            [30,6],
            [35,5.6],
            [120,25],
            [110,35],
            [120,30],
            [140,45]])
Y=np.array([1,1,1,0,0,0,0])
Y
```

```python
data=[45,10]
distance=np.array([np.sqrt(np.sum(x-data)**2) for x in X])
distance
```

```python
np.argsort(distance)  #argsort gives the index of the sorted array in ascending order
```

```python
d=np.argsort(distance)[:5] #gives the index of the 3 nearest neighbors
Y[d]
```
